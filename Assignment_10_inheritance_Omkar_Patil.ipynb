# Python OOP Assignment — Inheritance Examples

```python
# 1) Single Inheritance: Person -> Student
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no

# TODO:
# 1. Create object of Student.
obj=Student("Omkar",43)

# 2. Print name.
print(obj.name)

# 3. Print roll_no.
print(obj.roll_no)
```

```python
# 2) Single Inheritance: Animal -> Dog
class Animal:
    def sound(self):
        print("Some sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

# TODO:
# 1. Create Dog object.
obj=Dog()

# 2. Call sound() from parent class.
obj.sound()

# 3. Call bark() from child class.
obj.bark()
```

```python
# 3) Constructor Inheritance: Vehicle -> Car
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

# TODO:
# 1. Create Car object.
obj=Car("TATA","Sierra")

# 2. Print brand.
print(obj.brand)

# 3. Print model.
print(obj.model)
```

```python
# 4) Multilevel Inheritance: A -> B -> C
class A:
    def showA(self):
        print("This is A")

class B(A):
    def showB(self):
        print("This is B")

class C(B):
    def showC(self):
        print("This is C")

# TODO:
# 1. Create C object.
obj=C()

# 2. Call showA(), showB(), showC().
obj.showA()
obj.showB()
obj.showC()
```

```python
# 5) Multilevel Inheritance with methods: LivingThing -> Animal -> Dog
class LivingThing:
    def living(self):
        print("I am alive")

class Animal(LivingThing):
    def eat(self):
        print("Eating…")

class Dog(Animal):
    def bark(self):
        print("Barking…")

# TODO:
# 1. Create Dog object.
dog=Dog()

# 2. Call living(), eat(), bark().
dog.living()
dog.eat()
dog.bark()
```

```python
# 6) Multiple Inheritance: Father + Mother -> Child
class Father:
    def father_method(self):
        print("Father method")

class Mother:
    def mother_method(self):
        print("Mother method")

class Child(Father, Mother):
    def child_method(self):
        print("Child method")

# TODO:
# 1. Create Child object.
obj=Child()

# 2. Call father_method(), mother_method(), child_method().
obj.father_method()
obj.mother_method()
obj.child_method()
```

```python
# 7) Constructor Inheritance: Product -> Laptop
class Product:
    def __init__(self, name):
        self.name = name

class Laptop(Product):
    def __init__(self, name, ram):
        super().__init__(name)
        self.ram = ram

# TODO:
# 1. Create Laptop object.
HP=Laptop("Victus",16)

# 2. Print name.
print(HP.name)

# 3. Print ram.
print(HP.ram)
```

```python
# 8) Multilevel with __init__: Country -> State -> City
class Country:
    def __init__(self, cname):
        self.cname = cname

class State(Country):
    def __init__(self, cname, sname):
        super().__init__(cname)
        self.sname = sname

class City(State):
    def __init__(self, cname, sname, cityname):
        super().__init__(cname, sname)
        self.cityname = cityname

# TODO:
# 1. Create City object.
kop=City("India","Maharastra","Kolhapur")

# 2. Print cname, sname, cityname.
print(kop.cname,"-->",kop.sname,"-->",kop.cityname)
# print(kop.sname)
# print(kop.cityname)
```

```python
# 9) Multiple Inheritance with methods: Teacher + Player -> Student
class Teacher:
    def teach(self):
        print("Teaching…")

class Player:
    def play(self):
        print("Playing…")

class Student(Teacher, Player):
    def study(self):
        print("Studying…")

# TODO:
# 1. Create Student object.
o=Student()

# 2. Call teach(), play(), study().
o.teach()
o.play() 
o.study()
```

```python
# 10) Constructor Inheritance: Shape -> Rectangle
class Shape:
    def __init__(self, color):
        self.color = color

class Rectangle(Shape):
    def __init__(self, color, length, width):
        super().__init__(color)
        self.length = length
        self.width = width

# TODO:
# 1. Create Rectangle object.
obj=Rectangle("Red",50,25)

# 2. Print color, length, width.
print(obj.color,"\n",obj.length,"\n",obj.width)
```

```python
# Data input

orders = [
    {
        "id": 1,
        "student": {"id": "S001", "name": "Aarav"},
        "items": [
            {"name": "Sandwich", "price": 50.0, "qty": 2},
            {"name": "Juice", "price": 30.0, "qty": 1},
        ],
        "payment_method": "Cash"
    },
    {
        "id": 2,
        "student": {"id": "S002", "name": "Neha"},
        "items": [
            {"name": "Pasta", "price": 80.0, "qty": 1},
            {"name": "Juice", "price": 30.0, "qty": 2},
        ],
        "payment_method": "Card"
    },
    {
        "id": 3,
        "student": {"id": "S001", "name": "Raj"},
        "items": [
            {"name": "Sandwich", "price": 50.0, "qty": 1},
            {"name": "Cookie", "price": 20.0, "qty": 3},
        ],
        "payment_method": "Cash"
    }
]
```

### 1)  Total price


```python
## Without class Approach

# For each order add key "subtotal"
# Subtotal = sum of price * qty for all items.
# print "student_id", "student_name" and "subtotal"

for item in orders:
     item.update({"subtotal":0})
        
for order in orders:
    total = 0
    for item in order["items"]:
        total += item["price"] * item["qty"]
    order["subtotal"] = total 
    
print(orders)
print()     
    
for info in orders:
    print(f"ID:{info["student"]["id"]} | Name:{info["student"]["name"]} |Total price:{info["subtotal"]}")
```

```python
## with class Approach

# For each order add key "subtotal"
# Subtotal = sum of price * qty for all items.
# print "student_id", "student_name" and "subtotal"

class Order():
    def __init__(self,data):
        self.data=data
        self.subtotal = 0

    def cal_subtotal(self):
        for item in self.data["items"]:
            self.subtotal +=item["price"] * item["qty"] 
        self.data["subtotal"] = self.subtotal
        
    def display_info(self):
        print(f"ID: {self.data["student"]['id']} | Name: {self.data["student"]['name']} | Total price: {self.subtotal}")
        
        
for data in orders:
    obj=Order(data)
    obj.cal_subtotal()
    obj.display_info()

```

### 2)  Discount

```python
# Add more key "discount_price"
# Discount:
# If subtotal ≥ ₹120, apply 5% discount.
# Else, Original subtotal.
# print "student_id", "student_name" and "discount_price"


class discount_on_orders:
    
    def __init__(self,data):
        self.data=data
        self.discount=0
        
    def calculate_discount(self):
            if self.data["subtotal"]>=120:
                    self.discount=self.data["subtotal"]*0.05
                    self.data["Discount_price"]=self.discount       
            else:
                self.data["Discount_price"]=0
                
    def display_info(self):
        print(f"ID: {self.data["student"]['id']} | Name: {self.data["student"]['name']} | discount price: {self.discount}")
        
        
for info in orders:
    dis=discount_on_orders(info)
    dis.calculate_discount()
    dis.display_info()
```

### 3) Tax

```python
# Total tax applied  "tax"
# Tax: Apply 5% tax on the amount after discount.
# example subtotal is 100 then 5 is tax
#  print "student_id", "student_name" and "subtotal", "tax"

class total_tax():
    
    def __init__(self,data):
        self.data=data
        self.tax=0
        
    def claculate_tax(self):
        self.tax=self.data["subtotal"]*0.05
        self.data["Total_Tax"]=self.tax
        
    def display_info(self):
        print(f"Id:{self.data["student"]["id"]} | Name:{self.data["student"]["name"]} | Total tax:{self.tax}")
        
for info in orders:
    tax=total_tax(info)
    tax.claculate_tax()
    tax.display_info()
```

### 4)  Revenue

```python
# Total revenue for the day (sum of final totals).
# print revenue


class Total_revenue():
    
    def __init__(self,data):
        self.data=data
        
    def calculate_revenue(self):
        global revenue
        revenue +=self.data["subtotal"]+self.data["Total_Tax"]-self.data["Discount_price"]
       
        
revenue=0

for info in orders:
    rev=Total_revenue(info)
    rev.calculate_revenue()
    
    
print("Total Revenue:",revenue)
        
```

```python
orders = [
    {
        "id": 1,
        "student": {"id": "S001", "name": "Aarav"},
        "items": [
            {"name": "Sandwich", "price": 50.0, "qty": 2},
            {"name": "Juice", "price": 30.0, "qty": 1},
        ],
        "payment_method": "Cash"
    },
    {
        "id": 2,
        "student": {"id": "S002", "name": "Neha"},
        "items": [
            {"name": "Pasta", "price": 80.0, "qty": 1},
            {"name": "Juice", "price": 30.0, "qty": 2},
        ],
        "payment_method": "Card"
    },
    {
        "id": 3,
        "student": {"id": "S001", "name": "Raj"},
        "items": [
            {"name": "Sandwich", "price": 50.0, "qty": 1},
            {"name": "Cookie", "price": 20.0, "qty": 3},
        ],
        "payment_method": "Cash"
    }
]

""" Use Class Based Approach """

# For each order add key "subtotal"
# Subtotal = sum of price * qty for all items.
# print "student_id", "student_name" and "subtotal"

# Add more key "discount_price"
# Discount:
# If subtotal ≥ ₹120, apply 5% discount.
# Else, Original subtotal.
# print "student_id", "student_name" and "discount_price"

# Total tax applied  "tax"
# Tax: Apply 5% tax on the amount after discount.
# example subtotal is 100 then 5 is tax
#  print "student_id", "student_name" and "subtotal", "tax"

# Total revenue for the day (sum of final totals).
# print revenue
```

```python
# subtotal
class Order:
    def __init__(self,data):
        self.data=data
        self.subtotal = 0

    def cal_subtotal(self):
        for item in self.data["items"]:
            self.subtotal +=item["price"] * item["qty"] 
        self.data["subtotal"] = self.subtotal
        
    def display_info_subtotal(self):
        print(f"ID: {self.data["student"]['id']} | Name: {self.data["student"]['name']} | Total price: {self.subtotal}")
        
# discount       
class discount_on_orders(Order):
    
    def __init__(self,data):
        super().__init__(data)
        self.data=data
        self.discount=0
        
    def calculate_discount(self):
            if self.data["subtotal"]>=120:
                    self.discount=self.data["subtotal"]*0.05
                    self.data["Discount_price"]=self.discount       
            else:
                self.data["Discount_price"]=0
                
    def display_info_discount(self):
        print(f"ID: {self.data["student"]['id']} | Name: {self.data["student"]['name']} | discount price: {self.discount}")
        
# Tax
class total_tax(discount_on_orders):
    
    def __init__(self,data):
        super().__init__(data)
        self.data=data
        self.tax=0
        
    def claculate_tax(self):
        self.tax=self.data["subtotal"]*0.05
        self.data["Total_Tax"]=self.tax
        
    def display_info_tax(self):
        print(f"Id:{self.data["student"]["id"]} | Name:{self.data["student"]["name"]} | Total tax:{self.tax}")
  
  
# total revenue        
class Total_revenue(total_tax):
    
    def __init__(self,data):
        super().__init__(data)
        self.data=data
        
    def calculate_revenue(self):
        global revenue
        revenue +=self.data["subtotal"]+self.data["Total_Tax"]-self.data["Discount_price"]
 
 
revenue=0        
for info in orders:
    all=Total_revenue(info)
    all.cal_subtotal()
    all.display_info_subtotal()
    all=Total_revenue(info)
    all.calculate_discount()
    all.display_info_discount()
    all=Total_revenue(info)
    all.claculate_tax()
    all.display_info_tax()
    all.calculate_revenue()
    print()
    
print("Total revenue:",revenue)


```
