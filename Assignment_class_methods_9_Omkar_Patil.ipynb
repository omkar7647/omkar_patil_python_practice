## Q1: Create a class called Student

Requirements:
- Class variable: `school_name = "ABC International School"`
- Constructor: takes `name` and `marks`
- Instance method: `display_details()`
- Class method: `change_school()`
- Static method: `is_pass(marks)` returns "Pass" if marks ≥ 35 else "Fail"


```python
class student:
    school_name="ABC International School"                   #class variable
    
    def __init__(self,name,marks):                            #constructor
        self.name=name
        self.marks=marks
          
    def display_details(self):                                 #instance method
        print(f"Name:{self.name} | Marks:{self.marks}")
     
    @classmethod   
    def change_school(cls):                                    #class method
        cls.school_name="KIT International School"
        print(f"Changed school name:{cls.school_name}")
    
    @staticmethod
    def is_pass(marks):                                        #static method
        if marks>=35:
            print("Pass")
        else:
            print("Fail")
            

obj=student("omkar",40)

obj.display_details()
obj.change_school()
obj.is_pass(obj.marks)
```

## Q2: Create a class Calculator

Requirements:
- Class variable: `count_operations = 0`
- Instance method: `add(a, b)`
- Class method: `increment_count()`
- Static method: `info()`


```python
class calculator:
    count_operations=0                               #class variable
    
    def add(self,a,b):                                     #instance method
       print("Addition:",a+b)

        
    @classmethod 
    def increment_count(cls):                             #class method
        cls.count_operations+=5
        print("count operation:",cls.count_operations)
    
    @staticmethod
    def info():                                           #static method
        print("All information printed")
        
ob=calculator()
ob.add(10,20)
ob.increment_count()
ob.info()
```

## Q3: Write a class BankAccount

Requirements:
- Class variable: `bank_name = "SBI"`
- Constructor: `name`, `balance`
- Instance method: `deposit(amount)`
- Class method: `change_bank(new_name)`
- Static method: `validate_amount(amount)`


```python
class bankaccount:
    bank_name="SBI"                           #class variable
    
    def __init__(self,name,balence):           # constructor
        self.name=name
        self.balence=balence
        
    def deposit(self,amount):                     # instance method
        dep_amt=amount
        print("\nYour deposit amount is:",dep_amt)
        f_amount=self.balence+dep_amt
        print("Bank balence is:",f_amount)
        return f_amount
      
    @classmethod    
    def change_bank(cls,new_name):                      #class method
        cls.bank_name=new_name
        print("\nNew bank name:",cls.bank_name)
        
    @staticmethod
    def validate_amount(amount):                         #static method
        print("\nvalidate amount:",amount)
        
        
        
obj=bankaccount("omkar",15000)
s=obj.deposit(10000)
obj.change_bank("BOI")
obj.validate_amount(s)
    
```
