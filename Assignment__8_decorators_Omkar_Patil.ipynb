**Q22.** Explain (in a comment) why `convert_temperature()` should be an instance method if it depends on an object’s `scale` (C or F). Optionally, show a small example.

```python
# TODO: Write your solution for Q22 below

# convert_temperature() should be an instance method because the conversion depends on each object's 'scale' (C or F).  
# Each object can have a different scale, so we need 'self.scale' to know how THAT specific object converts temperatures.

class temp:
    def __init__(self, scale):   # scale = "C" or "F"
        self.scale = scale

    def convert_temperature(self,value):   # instance method
        if self.scale == "C":
            print((value*9/5)+32)  
        else:
            print((value-32)*5/9)

c_to_f = temp("C")
c_to_f.convert_temperature(25)

f_to_c = temp("F")
f_to_c.convert_temperature(77)

        

        
```

```python

class temp:
    def __init__(self, scale,value):   # scale = "C" or "F"         # always instance method
        self.scale = scale
        self.value=value
        self.convert_temperature()

    def convert_temperature(self):  
        if self.scale == "C":
            print((self.value*9/5)+32,"F")  
        else:
            print((self.value-32)*5/9,"C")

c_to_f = temp("C",25)
f_to_c = temp("F",77)
```

```python
# TODO: Write your solution for Q24 below

class date:
     
    def __init__(self,str_date):
        self.date=str_date.split() 
        self.from_string()
        
    dic={ "jan": 1, "feb": 2, "mar": 3, "apr": 4,"may": 5, "jun": 6, "jul": 7, "aug": 8,"sep": 9, "oct": 10, "nov": 11, "dec": 12}

    def from_string(self):
        month = self.date[1].lower()
        
        if month in self.dic:              
            self.date[1] = self.dic[month] 
            print(self.date[0],"-",self.date[1],"-", self.date[2])
        else:
            print("Wrong month entered")
         
        
obj=date("1 FEB 2025")
obj=date("1 dec 2003")
```

```python
#### method 2

class Date:
    
    dic = { 
        "jan": 1, "feb": 2, "mar": 3, "apr": 4,
        "may": 5, "jun": 6, "jul": 7, "aug": 8,
        "sep": 9, "oct": 10, "nov": 11, "dec": 12
    }

    def __init__(self, str_date):
        self.date = str_date.split()
        self.from_string()   

    @classmethod
    def convert_month(cls, month_str):
        month_str = month_str.lower()
        if month_str in cls.dic:
            return cls.dic[month_str]
        else:
            return None

    def from_string(self):
        month = self.convert_month(self.date[1])  
        if month:
            self.date[1] = month
            print(self.date[0], "-", self.date[1], "-", self.date[2])
        else:
            print("Wrong month entered")


obj = Date("1 FEB 2025")
obj = Date("1 nov 2011")

```

## G. Decorators (27–30)


**Q27.** Write a simple decorator that prints `"Function started"` before running the function. Test it on a small function.

```python
# TODO: Write your solution for Q27 below

def simple_decorator(fun):
    def inner(*arg,**kwarg):
        print("Functions started")
        fun(*arg,**kwarg)
    return inner
        
        

@simple_decorator
def print_():
        print("function task completed")
    
print_()
```

**Q28.** Create a decorator that prints `"Function ended"` after running the function. Test it.

```python
# TODO: Write your solution for Q28 below

def simple_decorator(fun):
    def inner(*arg,**kwarg):
        fun(*arg,**kwarg)
        print("Functions ended")
    return inner
        
        

@simple_decorator
def print_():
        print("function task completed")
        
print_()
```

**Q29.** Make a decorator that prints the **name** of the function being called (use `func.__name__`).

```python
# TODO: Write your solution for Q29 below

def function_name(funct):
    def inner(*arg,**kwarg):
        funct(*arg,**kwarg)
        print("Function Name:",funct.__name__)
    return inner
        
        
@function_name
def add(a,b):
    print ("sum =",a+b)
    
add(5,2)
```

**Q30.** Combine two decorators: one prints `"Start"`, another prints `"End"` around a function. Check the order of messages.

```python
# TODO: Write your solution for Q30 below

def start(funct):
    def inner(*arg,**kwarg):
        print("Stared:",funct.__name__)
        funct(*arg,**kwarg)
        
    return inner

def end(funct):
    def inner(*arg,**kwarg):
        funct(*arg,**kwarg)
        print("Ended:",funct.__name__)
    return inner


@start
@end
def testing():
    print("function testing completed")
    
testing()
```
